```python
import pandas as pd
from pathlib import Path

print("FreshCart Retail Analytics")
print("Data Exploration Started!")
```

```python
# Project directory
PROJECT_DIR = Path.cwd().parent

# Raw dataset directory
RAW_DATA_DIR = PROJECT_DIR / "data" / "raw"

print("Project directory:", PROJECT_DIR)
print("Raw data directory:", RAW_DATA_DIR)

# List all CSV files
csv_files = list(RAW_DATA_DIR.glob("*.csv"))

print("\nAvailable datasets:")

for file in csv_files:
    print(file.name)
```

```python
files = [
    "retail_inventory_ml_apl.csv",
    "BigBasket Products.csv",
    "retail_store_inventory.csv",
    "retail_sales_ml_apl.csv"
]

for file in files:
    df = pd.read_csv(RAW_DATA_DIR / file)

    print(f"\n{'='*60}")
    print(f"FILE: {file}")
    print(f"Shape: {df.shape}")
    print("\nColumns:")
    print(df.columns.tolist())
    print("\nFirst 3 rows:")
    display(df.head(3))
    print("\nMissing values:")
    display(df.isnull().sum())
```

```python
for file in files:
    df = pd.read_csv(RAW_DATA_DIR / file)

    print(f"\n{'='*60}")
    print(f"DATASET: {file}")

    print("\nDATA TYPES:")
    print(df.dtypes)

    print("\nDUPLICATE ROWS:")
    print(df.duplicated().sum())

    print("\nNUMERICAL SUMMARY:")
    display(df.describe().round(2))

    print("\nUNIQUE VALUES:")
    print(df.nunique())
```

```python
# 1. Inventory anomalies
inventory = pd.read_csv(RAW_DATA_DIR / "retail_inventory_ml_apl.csv")

print("INVENTORY ANOMALIES")
print("Negative stock:", (inventory["Qty on hand"] < 0).sum())
print("Negative stock value:", (inventory["Stocks Selling Amount"] < 0).sum())
print("Negative stock cost:", (inventory["Cost of Stocks"] < 0).sum())
print("Zero selling price:", (inventory["Stock Unit Selling Price"] == 0).sum())
print("Negative unit cost:", (inventory["Stock Unit Cost Price"] < 0).sum())


# 2. Demand forecasting anomalies
demand = pd.read_csv(RAW_DATA_DIR / "retail_store_inventory.csv")

print("\nDEMAND DATA ANOMALIES")
print("Negative forecasts:", (demand["Demand Forecast"] < 0).sum())
print("Zero units sold:", (demand["Units Sold"] == 0).sum())
print("Negative inventory:", (demand["Inventory Level"] < 0).sum())


# 3. Sales anomalies
sales = pd.read_csv(RAW_DATA_DIR / "retail_sales_ml_apl.csv")

print("\nSALES DATA ANOMALIES")
print("Return records:", (sales["Is Return"] == 1).sum())
print("Negative quantity:", (sales["Qty Sold"] < 0).sum())
print("Negative sales amount:", (sales["Sales Amount"] < 0).sum())
print("Negative COGS:", (sales["Cogs"] < 0).sum())

print("\nReturn breakdown:")
display(sales.groupby("Is Return").agg(
    records=("Product No", "count"),
    quantity=("Qty Sold", "sum"),
    sales=("Sales Amount", "sum"),
    cogs=("Cogs", "sum")
))
```

```python
# Negative inventory stock records
print("NEGATIVE STOCK RECORDS")
display(
    inventory[inventory["Qty on hand"] < 0][
        ["Product No", "Store", "Stock Status",
         "Qty on hand", "Stocks Selling Amount",
         "Cost of Stocks", "Stock Unit Selling Price",
         "Stock Unit Cost Price"]
    ].head(10)
)

# Stock status distribution
print("\nSTOCK STATUS DISTRIBUTION")
display(inventory["Stock Status"].value_counts())

# Negative stock grouped by stock status
print("\nNEGATIVE STOCK BY STATUS")
display(
    inventory.assign(
        Negative_Stock=inventory["Qty on hand"] < 0
    ).groupby("Stock Status")["Negative_Stock"].agg(
        ["count", "sum"]
    )
)

# Zero selling prices
print("\nZERO SELLING PRICE RECORDS")
display(
    inventory[inventory["Stock Unit Selling Price"] == 0][
        ["Product No", "Product Description", "Store",
         "Qty on hand", "Stock Unit Selling Price",
         "Stock Unit Cost Price"]
    ].head(10)
)

# Negative unit cost
print("\nNEGATIVE UNIT COST RECORDS")
display(
    inventory[inventory["Stock Unit Cost Price"] < 0].head(10)
)
```

```python
# Calculate expected inventory values
inventory["Expected_Selling_Value"] = (
    inventory["Qty on hand"] *
    inventory["Stock Unit Selling Price"]
)

inventory["Expected_Cost_Value"] = (
    inventory["Qty on hand"] *
    inventory["Stock Unit Cost Price"]
)

# Compare expected and reported values
inventory["Selling_Value_Difference"] = (
    inventory["Stocks Selling Amount"] -
    inventory["Expected_Selling_Value"]
)

inventory["Cost_Value_Difference"] = (
    inventory["Cost of Stocks"] -
    inventory["Expected_Cost_Value"]
)

print("SELLING VALUE VALIDATION")
print(
    "Mismatches:",
    (inventory["Selling_Value_Difference"].abs() > 0.02).sum()
)

print("\nCOST VALUE VALIDATION")
print(
    "Mismatches:",
    (inventory["Cost_Value_Difference"].abs() > 0.02).sum()
)

print("\nSAMPLE MISMATCHES")
display(
    inventory[
        (inventory["Selling_Value_Difference"].abs() > 0.02) |
        (inventory["Cost_Value_Difference"].abs() > 0.02)
    ][[
        "Product No",
        "Qty on hand",
        "Stocks Selling Amount",
        "Expected_Selling_Value",
        "Cost of Stocks",
        "Expected_Cost_Value"
    ]].head(10)
)
```

```python
# Identify cost valuation mismatches
cost_mismatches = inventory[
    inventory["Cost_Value_Difference"].abs() > 0.02
].copy()

# Split into two groups
zero_qty_mismatches = cost_mismatches[
    cost_mismatches["Qty on hand"] == 0
]

nonzero_qty_mismatches = cost_mismatches[
    cost_mismatches["Qty on hand"] != 0
]

print("TOTAL COST MISMATCHES:", len(cost_mismatches))

print("\nZERO-QUANTITY MISMATCHES:", len(zero_qty_mismatches))
print("NON-ZERO-QUANTITY MISMATCHES:", len(nonzero_qty_mismatches))

print("\nNON-ZERO QUANTITY EXAMPLES:")
display(
    nonzero_qty_mismatches[
        [
            "Product No",
            "Store",
            "Qty on hand",
            "Stock Unit Cost Price",
            "Cost of Stocks",
            "Expected_Cost_Value",
            "Cost_Value_Difference"
        ]
    ].head(15)
)

print("\nMISMATCHES BY STOCK STATUS:")
display(cost_mismatches["Stock Status"].value_counts())
```

```python
quality_report = pd.DataFrame([
    {
        "Dataset": "retail_inventory_ml_apl",
        "Check": "Duplicate rows",
        "Issue_Count": int(inventory.duplicated().sum()),
        "Decision": "No action required"
    },
    {
        "Dataset": "retail_inventory_ml_apl",
        "Check": "Negative stock quantity",
        "Issue_Count": int((inventory["Qty on hand"] < 0).sum()),
        "Decision": "Flag for investigation"
    },
    {
        "Dataset": "retail_inventory_ml_apl",
        "Check": "Negative unit cost",
        "Issue_Count": int((inventory["Stock Unit Cost Price"] < 0).sum()),
        "Decision": "Flag for investigation"
    },
    {
        "Dataset": "retail_inventory_ml_apl",
        "Check": "Zero selling price",
        "Issue_Count": int((inventory["Stock Unit Selling Price"] == 0).sum()),
        "Decision": "Validate separately"
    },
    {
        "Dataset": "retail_inventory_ml_apl",
        "Check": "Cost valuation mismatch",
        "Issue_Count": len(cost_mismatches),
        "Decision": "Zero-quantity records only; preserve"
    },
    {
        "Dataset": "retail_store_inventory",
        "Check": "Negative demand forecast",
        "Issue_Count": int((demand["Demand Forecast"] < 0).sum()),
        "Decision": "Flag before modelling"
    },
    {
        "Dataset": "retail_sales_ml_apl",
        "Check": "Return transactions",
        "Issue_Count": int((sales["Is Return"] == 1).sum()),
        "Decision": "Preserve as returns"
    }
])

display(quality_report)

# Save the report
quality_report.to_csv(
    "../reports/data_quality_report.csv",
    index=False
)

print("Data quality report saved successfully!")
```
